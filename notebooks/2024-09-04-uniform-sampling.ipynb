```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
%config InlineBackend.figure_format='retina'

plt.rcParams['text.usetex'] = True
plt.rcParams['xtick.labelsize'] = 12
plt.rcParams['ytick.labelsize'] = 12
```

```python
pts_x = np.random.rand(500) * 2 + 1
pts_y = np.random.rand(500) * 3 + 2
```

```python
fig, ax = plt.subplots(1, figsize=(4, 4))
plt.scatter(pts_x, pts_y)
plt.savefig('../images/uniform_square.png')
```

```python
pts_r = np.random.random(500) * 4
pts_theta = np.random.random(500) * 2 * np.pi

pts_x = pts_r * np.sin(pts_theta)
pts_y = pts_r * np.cos(pts_theta)

fig, ax = plt.subplots(1, figsize=(4, 4))
plt.scatter(pts_x, pts_y, s=2)
plt.savefig('../images/uniform_circle_wrong.png')
```

```python
pts_r_prime = np.random.random(500) * 16
pts_r = np.sqrt(2 * pts_r_prime)
pts_theta = np.random.random(500) * 2 * np.pi

pts_x = pts_r * np.sin(pts_theta)
pts_y = pts_r * np.cos(pts_theta)

fig, ax = plt.subplots(1, figsize=(4, 4))
plt.scatter(pts_x, pts_y, s=2)
plt.savefig('../images/uniform_circle_correct.png')
```
